```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Load the dataset
df = pd.read_csv("../data/ecommerce_sales_data.csv")

# Display first 5 rows
df.head()
```

```python
# Check number of rows and columns
print("Dataset Shape:", df.shape)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
# Display all column names
print("Column Names:")
print(df.columns.tolist())
```

```python
# Display dataset information
df.info()
```

```python
# Generate basic statistical information
df.describe()
```

```python
# Convert Order Date into datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Find the starting and ending dates
print("Starting Date:", df['Order Date'].min())
print("Ending Date:", df['Order Date'].max())
```

```python
# Count unique products, categories, and regions

print("Unique Products:", df['Product Name'].nunique())

print("Unique Categories:", df['Category'].nunique())

print("Unique Regions:", df['Region'].nunique())
```

```python
# Display a simple dataset summary

print("DATASET SUMMARY")
print("--------------------------")

print("Total Records:", len(df))
print("Total Columns:", len(df.columns))
print("Total Products:", df['Product Name'].nunique())
print("Total Categories:", df['Category'].nunique())
print("Total Regions:", df['Region'].nunique())

print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())
print("Total Quantity:", df['Quantity'].sum())
```
